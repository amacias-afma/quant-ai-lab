```python
import os
import sys
project_src = os.path.abspath('../outputs')
if project_src not in sys.path:
    sys.path.insert(0, project_src)
```

```python
import pandas as pd

df_results = pd.read_csv('../outputs/batch_summary.csv')
df_results
```

```python
fail_str = '❌ FAIL'
pass_str = '✅ PASS'

df_results['status_aux'] = 1
df_results.loc[df_results['Status'] == fail_str, 'status_aux'] = 0

df_results_pt = df_results.pivot_table(columns='Model', index='Ticker', values='status_aux')

```

```python
df_results_pt = df_results_pt.astype(str)

df_results_pt[df_results_pt == '1.0'] = pass_str
df_results_pt[df_results_pt == '0.0'] = fail_str

df_results_pt
```
